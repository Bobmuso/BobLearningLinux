Applications

The kernal is responsible for resource management, abstraction as well as multitasking 
This helps the machine run applications, such as Server application desktop application and other tools.

Server Applications:

web servers like Apache, word Press and Nginx. the Apache HTTPD serves requests and the users view them through HTTPS.

Private Cloud Servers, Ownlcloud in 2010 and Nextcloud in 2016 by Frank Karlitschek

Database Servers, MariaDB a fork of mySQL and more popular Firebird and PostgreSQL they both use SQL

Email Servers, Dovecot and Cyrus IMAP are POP/IMAP Server this is the Post Office Protocol (POP) and Internet Message Access Protocol (IMAP)

File Sharing: Samba allows Linux to transfer files with windows. Netatalk allows Mac transfer using the Network File System (NFS), which runs on the kernel

Lightweight Directory Access Protocol (LDAP) is the Domain Name System (DNS) alternative for linux infrastructure the name in OpenLDAP

Desktop applications:

Email: Thunderbird, Evolution and KMail.
Creative:  Blender, GIMP (GNU Image Manipulation Program), and Audacity.
Productivity: LibreOffice is a fork of the OpenOffice.
